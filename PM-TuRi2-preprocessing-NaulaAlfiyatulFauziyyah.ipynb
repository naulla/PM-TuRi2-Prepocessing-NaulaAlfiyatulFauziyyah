**C. Langkah Praktikum**

```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
df
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
print(X)
print(y)
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
print(X)
print(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)
print(X_train)
print(y_train)
print(X_test)
print(y_test)
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num]) # cegah data leakage
```

Jawaban: Label encoding digunakan untuk data yang memiliki urutan/ordinal dan hasilnya berupa satu kolom angka. Sedangkan one-hot encoding digunakan untuk data yang tidak memiliki urutan/nominal dan hasilnya berupa beberapa kolom dengan nilai 0 ataupun 1.  

3. Apa perbedaan normalisasi dan standardisasi?

Jawaban: Perbedaannya terletak pada hasil skala datanya. Normalisasi adalah proses mengubah skala data ke rentang tetap dan berguna ketika data membutuhkan batasan nilai yang jelas, tetapi normalisasi sensitif terhadap outlier. Sedangkan standardisasi adalah proses mengubah data sehingga memiliki mean/rata-rata 0 dan standar deviasi 1 dan hasilnya tidak memiliki batas rentang tertentu dan lebih tahan terhadap outlier.
